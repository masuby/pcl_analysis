```python
import xlwings as xw

def copy_sheet_and_create_table():
    master_file = r"C:\Users\Daniel\Desktop\code\CRM\Email\cs_crm_20_nov_2025.xlsx"
    processed_file = r"C:\Users\Daniel\Desktop\code\CRM\Email\cs_crm_19_again_nov_2025.xlsx"
    
    # Run Excel in the background (headless)
    app = xw.App(visible=False)
    
    try:
        # Open workbooks
        master_wb = app.books.open(master_file)
        processed_wb = app.books.open(processed_file)
        
        # Get user_data sheets
        master_ws_user = master_wb.sheets['user_data']
        processed_ws_user = processed_wb.sheets['user_data']

        # Get activity_data sheets
        master_ws_activity = master_wb.sheets['activity_data']
        processed_ws_activity = processed_wb.sheets['activity_data']

        # Get agent_activity sheets
        master_ws_agent = master_wb.sheets['agent_activity']
        processed_ws_agent = processed_wb.sheets['agent_activity']

        # Get Today's activity_data sheets
        master_ws_today = master_wb.sheets["Today's activity_data"]
        processed_ws_today = processed_wb.sheets["Today's activity_data"]

        # Get Lead_Report sheets
        master_ws_lead = master_wb.sheets["Lead_Report"]
        processed_ws_lead = processed_wb.sheets["Lead_Report"]
        
        print("Copying entire sheet...")
        
        # Clear the entire master sheet first
        master_ws_user.clear()
        master_ws_activity.clear()
        master_ws_agent.clear()
        master_ws_today.clear()
        master_ws_lead.clear()
    
        
        # Copy entire sheet from processed to master
        processed_ws_user.used_range.copy(master_ws_user.range('A1'))
        processed_ws_activity.used_range.copy(master_ws_activity.range('A1'))
        processed_ws_agent.used_range.copy(master_ws_agent.range('A1'))
        processed_ws_today.used_range.copy(master_ws_today.range('A1'))
        processed_ws_lead.used_range.copy(master_ws_lead.range('A1'))
        
        print("Sheet copied successfully!")
        

            
            
        
        # Save master file
        master_wb.save()
        print("Master file saved successfully!")
        
    except Exception as e:
        print(f"Error: {e}")
    finally:
        # Close workbooks and quit app
        try:
            processed_wb.close()
        except:
            pass
        try:
            master_wb.close()
        except:
            pass
        app.quit()

copy_sheet_and_create_table()
```
